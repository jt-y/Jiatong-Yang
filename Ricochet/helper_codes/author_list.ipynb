```python
import pandas as pd
```

```python
# ====== FILES ======
input_file = "Ricochet Collaboration Membership.xlsx"

# ====== LOAD ======
df = pd.read_excel(input_file)

# ====== FILTER AUTHORS ======
df = df[df["Al-Resonator Paper"].str.strip().str.lower() == "yes"]

# ====== CLEAN + FORMAT NAMES ======
def format_name(row):
    initials = row["Initial style"].replace(".", " ").strip()
    last = row["Last name"].strip()
    return f"{initials} {last}"

df["Formatted Name"] = df.apply(format_name, axis=1)

# ====== SORT ALPHABETICALLY BY LAST NAME ======
# Case insensitive sort by last name

df = df.sort_values(by="Last name", key=lambda x: x.str.lower())
```

```python
# ====== BUILD AFFILIATION MAP ======
affil_dict = {}
affil_list = []

for _, row in df.iterrows():
    for inst_col in ["Institution", "Institution 2", "Institution 3"]:
        if inst_col in df.columns:
            affil = row[inst_col]
            if pd.notna(affil) and affil.strip():  # Check for non-null and non-empty
                affil = affil.strip()
                if affil not in affil_dict:
                    affil_dict[affil] = len(affil_dict) + 1
                    affil_list.append(affil)

# ====== BUILD AUTHOR STRING ======
author_entries = []

for _, row in df.iterrows():
    name = row["Formatted Name"]
    affil_nums = []

    
    for inst_col in ["Institution", "Institution 2", "Institution 3"]:
        if inst_col in df.columns:
            affil = row[inst_col]
            if pd.notna(affil) and affil.strip():  # Check for non-null and non-empty
                affil = affil.strip()
                # sort affiliations by their assigned number
                affil_nums.append(affil_dict[affil])

    affil_nums.sort()
    affil_superscript = ", ".join(str(num) for num in affil_nums)
    author_entries.append(
        f"\\mbox{{{name}\\textsuperscript{{{affil_superscript}}}}}"
    )


author_line = "\\author{" + ", ".join(author_entries) + "}"

# ====== BUILD ADDRESS BLOCK ======
address_lines = []
for affil, idx in affil_dict.items():
    address_lines.append("\\address{$^{"+str(idx)+"}$"+affil+"}")

# ====== FINAL LATEX BLOCK ======
latex_output = author_line + "\n\n" + "\n".join(address_lines)

# ====== PRINT (copy-paste ready) ======
print(latex_output)
```
